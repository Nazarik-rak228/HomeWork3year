<h1>Задание 1</h1>

```python
import numpy as np ## импорт библиотеки 
```

```python
data_temperature = np.array([
54.1, 55.3, 53.8, 56.2, 54.9, 55.7, 54.4, 56.1, 55.0, 54.6,
95.0, 55.9, 54.2, 56.3, 55.1, 54.7, 55.4, 53.9, 56.0, 55.2,
54.5, 55.6, 54.3, 56.4, 55.0, 54.9, 55.8, 54.0, 56.1, 54.8,
55.5, 53.7, 56.2, 55.3, 54.4, 55.9, 54.1, 56.0, 55.1, 54.6,
55.7, 54.2, 56.3, 55.4, 53.8, 56.1, 54.9, 55.6, 54.3, 56.2,
99.0, 97.0, 55.0, 54.7, 55.8, 54.1, 56.4, 55.2, 54.5, 55.9,
54.0, 56.1, 55.3, 54.8, 55.6, 54.2, 56.3, 55.1, 54.6, 55.7,
100.0, 54.9, 55.4, 53.9, 56.0, 55.2, 54.4, 55.8, 54.1, 56.2,
55.0, 54.7, 55.5, 54.3, 56.1, 55.3, 54.6, 55.9, 54.0, 56.0,
55.1, 54.8, 55.6, 54.2, 56.3, 55.4, 53.8, 55.7, 54.9, 55.2
])
```

<h3>Среднее значение</h3>

<p>Среднее значение показывает, какая температура сервера наблюдается в среднем за все 100 замеров. Для его вычисления используется функция <code>np.mean()</code>.</p>

```python
print("Среднее значение:", np.mean(data_temperature))
```

<p>Средняя температура сервера составляет примерно 56.84 °C. На среднее значение повлияли несколько высоких температур, поэтому оно получилось выше основной группы значений около 54–56 °C.</p>

<h3>Медиана</h3>

<p>Медиана показывает центральное значение набора данных. Для её нахождения используется функция <code>np.median()</code>. Половина значений находится ниже медианы, а другая половина выше неё.</p>

```python
print("Медиана:", np.median(data_temperature))
```

<p>Медиана равна 55.2 °C. Она лучше показывает обычную температуру сервера, так как на неё практически не влияют отдельные очень высокие значения.</p>

<h3>Стандартное отклонение</h3>

<p>Стандартное отклонение показывает, насколько сильно значения температуры в среднем отклоняются от среднего значения. Так как данные 100 замеров являются выборкой из большего количества возможных замеров, используется <code>ddof=1</code>.</p>

```python
print("Стандартное отклонение:", np.std(data_temperature, ddof=1))
```

<p>Стандартное отклонение показывает заметное отклонение температуры от среднего значения. На его величину сильно повлияли четыре высоких значения температуры.</p>

<h3>Дисперсия</h3>

<p>Дисперсия показывает степень разброса значений относительно среднего значения. Она связана со стандартным отклонением, но измеряется в квадратных единицах. 

```python
print("Дисперсия:", np.var(data_temperature, ddof=1))
```

<p>Полученное значение дисперсии показывает, что наличие нескольких очень высоких температур значительно увеличивает общий разброс данных.</p>

<h3>Максимальное значение</h3>

<p>С помощью функции <code>np.max()</code> находится самое большое значение температуры в массиве.</p>

```python
print("Максимальное значение:", np.max(data_temperature))
```

<p>Максимальная температура сервера составила 100.0 °C. Это значительно выше основной группы значений и является одним из выбросов.</p>

<h3>Минимальное значение</h3>

<p>С помощью функции <code>np.min()</code> находится самое маленькое значение температуры в массиве.</p>

```python
print("Минимальное значение:", np.min(data_temperature))
```

<p>Минимальная температура составила 53.7 °C. Она находится рядом с остальными обычными значениями и не выглядит аномальной.</p>

<h3>Размах данных</h3>

<p>Размах показывает разницу между максимальным и минимальным значениями.</p>

```python
print("Размах:",np.ptp(data_temperature))
```

<p>Размах температуры составляет 46.3 °C. Такое большое значение получилось из-за высоких температур 
до 100 °C.</p>

<h3>Перцентили</h3>

<p>Перцентили показывают значение, ниже которого находится определённая часть всех наблюдений. В работе используются 25-й, 50-й, 75-й, 90-й и 99-й перцентили.</p>

```python
percentileis = [25, 50, 75, 90, 99]
for i in percentileis:
     print(f'{i} перцентель: {np.percentile(data_temperature,i)}')
    
```

<p>25-й перцентиль показывает границу, ниже которой находится примерно 25% наблюдений, 50-й перцентиль соответствует медиане, а 75-й перцентиль показывает границу для 75% наблюдений.</p>

<p>Высокое значение 99-го перцентиля связано с наличием нескольких температур около 100 °C.</p>

<h3>Первый и третий квартили</h3>

<p>Первый квартиль qw1 соответствует 25-му перцентилю, а третий квартиль qw3 соответствует 75-му перцентилю. Они разделяют упорядоченный набор данных на части.</p>

```python
qw1 = np.percentile(data_temperature, 25)
qw3 = np.percentile(data_temperature, 75)

print("Первый квартиль:", qw1)
print("Третий квартиль:", qw3)
```

<p>Для данного набора данных первый квартиль равен 54.575 °C, а третий квартиль равен 55.9 °C. Это показывает, что основная часть значений сосредоточена в достаточно узком диапазоне температур.</p>

<h3>Межквартильный размах</h3>

<p>Межквартильный размах показывает ширину центральных 50% данных. Он рассчитывается как разность между третьим и первым квартилями: <code>IQR = Q3 - Q1</code>.</p>

```python
IQR = qw3 - qw1
print("Межквартильный размах:", IQR) 
```

<p>Межквартильный размах равен 1.325 °C. Это небольшой разброс, поэтому основная часть температур находится близко друг к другу.</p>


<h3>Поиск выбросов</h3>

<p>Для поиска выбросов используется межквартильный размах. Значения, которые находятся ниже <code>Q1 - 1.5 × IQR</code> или выше <code>Q3 + 1.5 × IQR</code>, считаются выбросами.</p>

```python
low = qw1 - 1.5 * IQR
upp = qw3 + 1.5 * IQR

print("Нижняя граница:", low)
print("Верхняя граница:", upp)
```

```python

outliers = data_temperature[
    (data_temperature < low) |
    (data_temperature > upp)
]

print("Выбросы:", outliers)
print("Количество выбросов:", len(outliers))
```

<h4>Найденные выбросы</h4>

<p>В результате анализа были обнаружены четыре выброса: 95.0, 97.0, 99.0 и 100.0 °C. Все они значительно выше основной группы температур.</p>

<p>Такие значения могут быть связаны с кратковременным перегревом сервера, высокой нагрузкой или ошибкой измерительного датчика.</p>

<h2>Задание 2</h2>

```python
##  Массив A - данные 2 задания 
A = np.array([54.1, 55.3, 53.8, 56.2, 54.9, 55.7, 54.4, 56.1, 55.0, 54.6, 55.9, 54.2, 56.3, 55.1, 54.7])
```

<h3>Создание матрицы 3×5</h3>

<p>Массив <code>A</code> содержит 15 элементов. С помощью функции <code>reshape()</code> преобразуем его в матрицу размером 3×5.</p>


```python
matrix = A.reshape(3, 5)
```

```python
print("Матрица 3x5:")
print(matrix)
```

<p>Полученная матрица содержит 3 строки и 5 столбцов, при этом все исходные элементы массива сохранились.</p>

<h3>Размерность матрицы</h3>

<p>Свойство <code>shape</code> показывает размерность массива. Для полученной матрицы результат <code>(3, 5)</code> означает 3 строки и 5 столбцов.</p>

```python
print("Размерность матрицы:", matrix.shape)
```

<h3>Количество элементов</h3>

<p>Свойство <code>size</code> показывает общее количество элементов массива. В матрице 3×5 находится 15 элементов.</p>

```python
print("Количество элементов:", matrix.size)
```

<h3>Сумма по строкам</h3>

<p>С помощью <code>sum(axis=1)</code> вычисляется сумма элементов отдельно для каждой строки матрицы.</p>


```python
print("Сумма по строкам:", matrix.sum(axis=1))
```

<p>Таким образом, для каждой из трёх строк была получена отдельная сумма её элементов.</p>

<h3>Сумма по столбцам</h3>

<p>С помощью <code>sum(axis=0)</code> вычисляется сумма элементов отдельно для каждого столбца матрицы.</p>

```python
print("Сумма по столбцам:", matrix.sum(axis=0))
```

<p>Получены пять значений, соответствующих суммам элементов каждого из пяти столбцов.</p>

<h3>Транспонирование матрицы</h3>

<p>Транспонирование меняет местами строки и столбцы матрицы. Для этого используется свойство <code>T</code>.</p>

```python
print("Транспонированная матрица:")
print(matrix.T)
```

<p>Исходная матрица имела размерность 3×5, после транспонирования её размерность стала 5×3.</p>

<h3>Статистический анализ матрицы</h3>

<p>Для дальнейшего анализа все 15 элементов матрицы рассматриваются как единый набор данных. Для него вычисляются среднее значение, стандартное отклонение, дисперсия и квартили.</p>

```python
print("Среднее значение:", np.mean(matrix))
print("Стандартное отклонение:", np.std(matrix,ddof=1))
print("Дисперсия:", np.var(matrix, ddof=1))
```

```python
print("Первый квартиль:", np.percentile(matrix, 25))
print("Второй квартиль:", np.percentile(matrix, 50))
print("Третий квартиль:", np.percentile(matrix, 75))
```

<p>Среднее значение показывает среднюю температуру среди всех элементов матрицы. Стандартное отклонение и дисперсия показывают разброс этих значений, а квартили позволяют оценить распределение данных.</p>
