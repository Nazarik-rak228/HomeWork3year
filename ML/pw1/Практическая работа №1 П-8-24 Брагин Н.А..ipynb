<h2>3. Обработка датасета</h2>

<h3>3.1. Удаление признака</h3>

```python
print('Уникальных значений dataset_key:', df['dataset_key'].nunique())
print('Примеры dataset_key:', df['dataset_key'].unique()[:3].tolist())
df = df.drop(columns=['dataset_key'])
print('Столбцы после удаления:')
print(df.columns.tolist())
print('Размер таблицы:', df.shape)
```

<code>df.drop(columns=[...])</code> убирает указанные столбцы. Важно присвоить результат обратно в <code>df</code> (или использовать <code>inplace=True</code>), иначе исходная таблица не изменится.

Удалён <code>dataset_key</code> - это технический UUID исходного набора данных внутри GBIF. Он не описывает ни вид, ни место, ни время наблюдения и не несёт пользы ни для описательной статистики, ни для будущей модели. После удаления остаётся 13 признаков.

<h3>3.2. Замена и удаление значений</h3>

```python
print('Пропуски до замены:')
print(df[['month', 'individual_count', 'coordinate_uncertainty_m']].isna().sum())

df['month'] = df['month'].fillna(0)  # 0 = месяц не указан
df['individual_count'] = df['individual_count'].fillna(1)  # по умолчанию 1 особь - логично если наблюдение было, значит была хотя бы 1 особь
df['coordinate_uncertainty_m'] = df['coordinate_uncertainty_m'].fillna(-1)  # -1 = неизвестно


df['month'] = df['month'].astype(int)
df['individual_count'] = df['individual_count'].astype(int)

print()
print('Пропуски после замены:')
print(df[['month', 'individual_count', 'coordinate_uncertainty_m']].isna().sum())
print()
print('Значения month после замены (0 = не указан):')
print(df['month'].value_counts().sort_index())
```

На этом шаге ликвидируются пропуски методом <code>fillna</code>.

Логика замен:
<ul>
<li><code>month</code> → 0 — «месяц не указан» (0 не пересекается с реальными 1–12);</li>
<li><code>individual_count</code> → 1 - в большинстве наблюдений отмечается одна особь, и логически если было наблюдение - был минимум 1 объект наблюдения;</li>
<li><code>coordinate_uncertainty_m</code> → −1 - «точность неизвестна» (отрицательное число не путается с реальной погрешностью ≥ 0).</li>
</ul>

Удалять все строки с пропусками было бы неверно: потерялась бы значительная доля данных. Замена сохраняет наблюдения и явно помечает отсутствие информации. После заполнения <code>month</code> и <code>individual_count</code> приведены к целому типу через <code>astype(int)</code>.

```python
rows_before = df.shape[0]

mask_weak = (df['coordinate_uncertainty_m'] > 5000) & (df['month'] == 0)
print('Строк с большой неопределённостью и без месяца:', mask_weak.sum())

df = df.loc[~mask_weak].copy()
df = df.reset_index(drop=True)

print('Строк до удаления:', rows_before)
print('Строк после удаления:', df.shape[0])
print('Удалено строк:', rows_before - df.shape[0])
```

Часть строк можно удалить как малоинформативные.

Строится булева маска: неопределённость координат больше 5000 м <b>и</b> месяц не указан. Такие записи одновременно «размыты» в пространстве и во времени.

Оператор <code>~</code> инвертирует маску. <code>df.loc[~mask]</code> оставляет только «хорошие» строки. <code>.copy()</code> создаёт независимую копию (чтобы не словить предупреждения о SettingWithCopy). <code>reset_index(drop=True)</code> заново нумерует строки с нуля.

Таким образом наблюдения очищаются от неинформативных записей.

<h3>3.3. Сортировка значений по столбцу</h3>

```python
df = df.sort_values(by=['year', 'month'], ascending=[False, False])
df = df.reset_index(drop=True)
df.head(10)
```

<code>sort_values</code> упорядочивает строки по одному или нескольким столбцам.

Здесь сортировка идёт по паре <code>year</code>, <code>month</code> в порядке убывания: сначала самые свежие годы, внутри года — более поздние месяцы. После сортировки индекс «ломается», поэтому его сбрасывают через <code>reset_index(drop=True)</code>.

<code>head(10)</code> показывает десять самых новых записей — удобно для быстрой проверки, что порядок действительно изменился.

<h3>3.4. Удаление дубликатов</h3>

```python
print('Строк до удаления дубликатов:', df.shape[0])
print('Полных дубликатов:', df.duplicated().sum())
print('Повторов gbif_id:', df.duplicated(subset=['gbif_id']).sum())

df = df.drop_duplicates()
df = df.drop_duplicates(subset=['gbif_id'], keep='first')
df = df.reset_index(drop=True)

print('Строк после удаления дубликатов:', df.shape[0])
```

<code>drop_duplicates()</code> удаляет повторяющиеся строки.

Без аргументов дубликатом считается полное совпадение всех столбцов. С <code>subset=['gbif_id']</code> достаточно совпадения идентификатора. Параметр <code>keep='first'</code> оставляет первое вхождение и отбрасывает остальные.

В текущей выборке повторов не оказалось. 

```python
print('Итоговый размер очищенной таблицы:', df.shape)
print()
print(df.info())
print()
df.head()
```

После всех преобразований полезно ещё раз вызвать <code>info()</code> и <code>head()</code>. Так видно итоговый размер таблицы, отсутствие пропусков в обработанных полях и то, что строки действительно идут от новых наблюдений к более старым.

<h2>4. Выбор целевой переменной</h2>

В качестве целевой переменной берём категориальный признак <code>common_name</code> — обычное название вида.

<b>Смысл признака.</b>  
Это метка вида: European Wildcat, Lion, Ocelot, Eurasian Lynx и т.д. Значений немного (около полутора десятков), все они текстовые — классическая категориальная переменная.

<b>Почему не подходят другие столбцы.</b>
<ul>
<li><code>gbif_id</code> — чистый идентификатор. Новый id не несёт информации о виде, поэтому для обучения модели он бесполезен.</li>
<li><code>scientific_name</code> по сути дублирует <code>common_name</code> на латыни. Для интерпретации удобнее привычные английские названия.</li>
<li><code>country_code</code> и <code>license</code> — тоже категории, но отвечают на вопросы «где» и «под какой лицензией», а не «какой вид наблюдали».</li>
<li><code>basis_of_record</code> описывает способ сбора (человек / датчик / образец), а не объект наблюдения.</li>
<li>Числовые поля (<code>year</code>, координаты, число особей) логичнее использовать как входные признаки, а не как класс.</li>
</ul>

<b>Какую задачу это задаёт.</b>  
Если <code>common_name</code> считать целью, дальше можно ставить задачу многоклассовой классификации: по стране, координатам, году, типу записи и лицензии пытаться отличить, например, льва от рыси или оцелота от дикой кошки.

<b>Ограничение.</b>  
Классы несбалансированы: European Wildcat встречается заметно чаще остальных. При построении модели это нужно будет учитывать, но для этапа разведочного анализа выбор вида как целевой переменной остаётся наиболее естественным.
